```python
def flames_game():
    print("--- FLAMES Game ---")
    name1 = input("Enter the first name: ").lower().replace(" ", "")
    name2 = input("Enter the second name: ").lower().replace(" ", "")

    # Convert names to lists of characters for easier manipulation
    list_name1 = list(name1)
    list_name2 = list(name2)

    # Remove common characters
    for char1 in name1:
        if char1 in list_name2:
            list_name1.remove(char1)
            list_name2.remove(char1)

    # Calculate the total number of remaining characters
    remaining_chars_count = len(list_name1) + len(list_name2)

    flames_relations = ["Friends", "Lovers", "Affectionate", "Marriage", "Enemies", "Siblings"]

    if remaining_chars_count == 0:
        # If all characters are common, let's say they are best friends or deeply connected
        result = "Deeply Connected"
    else:
        # Use modulo to cycle through the FLAMES string
        # The formula (count % len) - 1 works well, handling the '0' case for the last element
        idx = (remaining_chars_count % len(flames_relations)) - 1
        result = flames_relations[idx]

    print(f"\nBased on the names, the relationship is: {result}")

# Run the FLAMES game
flames_game()
```

## Inheritance Example

This section demonstrates object-oriented programming inheritance in Python. We will define a `Vehicle` base class and a `Car` derived class.

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def display_info(self):
        return f"Brand: {self.brand}, Model: {self.model}"

class Car(Vehicle):
    def __init__(self, brand, model, year):
        super().__init__(brand, model) # Call the base class constructor
        self.year = year

    def display_car_info(self):
        # Access method from the base class
        base_info = super().display_info()
        return f"{base_info}, Year: {self.year}"

# Create an instance of the base class
vehicle1 = Vehicle("Generic", "Vehicle A")
print("Vehicle 1 Info:", vehicle1.display_info())

# Create an instance of the derived class
car1 = Car("Toyota", "Camry", 2022)
print("Car 1 Info:", car1.display_car_info())

# Demonstrate that Car objects also have Vehicle methods
print("Car 1 as Vehicle Info:", car1.display_info())
```

## Types of Inheritance in Python

Python supports several types of inheritance:

1.  **Single Inheritance**: A child class inherits from a single parent class. This is the most common and simplest form of inheritance.
    *   Example: `Car` inheriting from `Vehicle`.

2.  **Multiple Inheritance**: A child class inherits from multiple parent classes. This allows a class to combine features from several base classes.
    *   Example: A `FlyingCar` class inheriting from both `Car` and `Aircraft`.

3.  **Multilevel Inheritance**: A child class inherits from a parent class, which in turn inherits from another parent class. This creates a chain of inheritance.
    *   Example: `Car` inherits from `Vehicle`, and `SportsCar` inherits from `Car`.

4.  **Hierarchical Inheritance**: Multiple child classes inherit from a single parent class. Each child class can have its own distinct features.
    *   Example: `Car`, `Motorcycle`, and `Truck` all inheriting from `Vehicle`.

5.  **Hybrid Inheritance**: A combination of two or more of the above types of inheritance. It often involves a mix of single, multiple, or multilevel inheritance.

## Summary: Single vs. Multiple Inheritance

| Feature           | Single Inheritance                                 | Multiple Inheritance                               |
| :---------------- | :------------------------------------------------- | :------------------------------------------------- |
| **Definition**    | A class inherits from one base class.             | A class inherits from multiple base classes.       |
| **Simplicity**    | Simpler to understand and implement.              | More complex; can lead to ambiguity.               |
| **Hierarchy**     | Creates a linear hierarchy (parent-child).        | Creates a directed acyclic graph (DAG) hierarchy.  |
| **Diamond Problem** | Not applicable.                                   | Can lead to the "Diamond Problem" (ambiguity in method resolution order).
| **Code Reusability** | Reuses code from a single parent.                 | Reuses code from multiple parents.                 |
| **Use Case**      | Common for specialization and extending one class. | Used when a class needs to combine distinct functionalities from different sources.

### 1. Method Overriding (Runtime Polymorphism)

This occurs when a subclass provides a specific implementation for a method that is already defined in its superclass. The method in the subclass 'overrides' the method in the superclass. When you call that method on an object, Python determines which implementation to use at runtime based on the object's actual type.

```python
class Animal:
    def speak(self):
        return "Animal makes a sound"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

# Demonstrating polymorphism
animal = Animal()
dog = Dog()
cat = Cat()

print(animal.speak())
print(dog.speak())
print(cat.speak())
```

### 2. Method Overloading (Compile-time Polymorphism - Simulated in Python)

Method overloading means having multiple methods with the same name but different parameters within the *same* class. While many languages support this directly, Python does not have built-in method overloading in the traditional sense (it only remembers the last defined method). However, you can achieve similar behavior using default arguments, variable-length arguments (`*args`, `**kwargs`), or conditional logic inside a single method.

```python
class Calculator:
    def add(self, a, b=None):
        if b is not None:
            return a + b
        else:
            return a + a # Example: if only one arg, double it

    def multiply(self, *args):
        result = 1
        for num in args:
            result *= num
        return result

calc = Calculator()
print(f"Add with two args: {calc.add(5, 3)}")
print(f"Add with one arg: {calc.add(7)}")
print(f"Multiply with multiple args: {calc.multiply(1, 2, 3, 4)}")
```

### 3. Operator Overloading

Python allows you to define how operators (like `+`, `-`, `*`, `==`) behave for user-defined classes. This is done by implementing special methods (often called 'dunder methods' because they start and end with double underscores, like `__add__`, `__mul__`, `__eq__`). This makes custom objects work with operators in an intuitive way, similar to built-in types.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(2, 3)
v2 = Vector(5, 7)
v3 = v1 + v2  # Uses the __add__ method

print(v3)
```

```python
class Contact:
    def __init__(self):
        self.book = {}


class Details(Contact):

    def Create(self, name, number):

        if name in self.book:
            print("Contact already exists")

        elif number in self.book.values():
            print("Number already exists")

        else:
            self.book[name] = number
            print("Contact Added Successfully")

    def view(self):

        if len(self.book) == 0:
            print("No Contact Found")

        else:
            print("\n----- Contact List -----")
            for name, number in self.book.items():
                print(f"Name : {name}\tPhone : {number}")

    def modify(self, old_name, old_number, new_name, new_number):

        if old_name in self.book and self.book[old_name] == old_number:

            del self.book[old_name]
            self.book[new_name] = new_number

            print("Contact Updated Successfully")

        else:
            print("Contact Not Found")


obj = Details()

while True:

    print("\n===== CONTACT MANAGEMENT SYSTEM =====")
    print("1. Create Contact")
    print("2. View Contacts")
    print("3. Modify Contact")
    print("4. Exit")

    n = int(input("Enter Choice: "))

    if n == 1:

        name = input("Enter Name: ")
        number = input("Enter Phone Number: ")

        if len(number) != 10:
            print("Invalid Contact Number")

        else:
            obj.Create(name, number)

    elif n == 2:

        obj.view()

    elif n == 3:

        old_name = input("Enter Existing Name: ")
        old_number = input("Enter Existing Number: ")

        new_name = input("Enter New Name: ")
        new_number = input("Enter New Number: ")

        if len(new_number) != 10:
            print("Invalid Contact Number")

        else:
            obj.modify(
                old_name,
                old_number,
                new_name,
                new_number
            )

    elif n == 4:

        print("Thank You...")
        break

    else:

        print("Invalid Choice")
```
